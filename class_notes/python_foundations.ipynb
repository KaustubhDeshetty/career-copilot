```
# Assignment 3 : Employee Risk Assessment System

employees = {
    "Emp_01" : {'name' : 'Kaustubh', 'satisfaction':1.2, 'is_overtime':True,  'years_since_promotion':1, 'years_worked':3, 'monthly_salary':69000},
    "Emp_02" : {'name' : 'Siva',     'satisfaction':4.3, 'is_overtime':False, 'years_since_promotion':2, 'years_worked':7, 'monthly_salary':81000},
    "Emp_03" : {'name' : 'Yagna',    'satisfaction':3.6, 'is_overtime':True,  'years_since_promotion':5, 'years_worked':8, 'monthly_salary':45000},
    "Emp_04" : {'name' : 'Poojitha', 'satisfaction':2.8, 'is_overtime':True,  'years_since_promotion':3, 'years_worked':5, 'monthly_salary':75000},
    "Emp_05" : {'name' : 'Abhimanyu','satisfaction':4.7, 'is_overtime':False, 'years_since_promotion':1, 'years_worked':2, 'monthly_salary':89000},
}

# Report Header
print(f"{'Name':<20} : {'Salary':<10} : {'Salary Cat':<15} : {'Risk':<10} : {'Action'}")
print("-" * 80)

# Summary Counters
# Initialized before the loop to accumulate values across all employees
employees_at_risk = 0    # counts employees with HIGH or CRITICAL risk
total_cost        = 0    # total cost (3 months salary) if at-risk employees quit

# Main Processing Loop
# Iterates through each employee to calculate salary category, risk, and action
for eid, details in employees.items():
    if details['monthly_salary'] < 50000:
        salary_category = 'Below Market'              # underpaid relative to market
    elif details['monthly_salary'] < 80000:
        salary_category = 'Market'                    # competitive salary range
    else:
        salary_category = 'Above Market'              # above average compensation

    score = 0

    if details['satisfaction'] < 2.0:
        score += 40                                    # critically low satisfaction
    elif details['satisfaction'] < 3.0:
        score += 25                                    # below average satisfaction

    #risk favcotr
    if details['is_overtime']:              score += 20    # working overtime adds burn=ut
    if details['years_since_promotion'] >= 4: score += 20  # overdue for promotion
    if details['years_worked'] > 6:         score += 10    # long years worked without growth

    # Determine risk level based on total score using ternary operator
    level = "CRITICAL" if score >= 70 else "HIGH" if score >= 50 else "MEDIUM" if score >= 30 else "LOW"

    # Assign action based on risk level
    action = ("Annual review is fine"        if level == "LOW"      else
              "Schedule quarterly check-in"  if level == "MEDIUM"   else
              "Manager meeting this week"    if level == "HIGH"      else
              "URGENT: HR intervention needed")

    # Print Employee Row
    print(f"{details['name']:<20} : {details['monthly_salary']:<10} : {salary_category:<15} : {level:<10}: {action}")

    # Accumulate data for employees flagged as HIGH or CRITICAL risk
    if level == 'HIGH' or level == 'CRITICAL':
        employees_at_risk += 1
        total_cost += details['monthly_salary'] * 3    # 3 months salary per employee

#Summary Report
print("-" * 80)
print(f"Employees at HIGH or CRITICAL risk : {employees_at_risk}")
print(f"Total cost if they quit            : {total_cost}")
```

```
#Function: Write once, use many times
def calculate_risk_score(satisfaction,overtime,years_since_promotion,salary):
    '''
    Calculate attrition risk score for one employee.
    Returns: (score 0-100, risk level string)
    '''
    score = 0

    if satisfaction < 0.3:    score += 40
    elif satisfaction < 0.5:  score +=25
    if overtime:              score +=20
    if years_since_promotion >= 4: score += 20
    if salary <  50000:        score += 15

    if score >= 70:     level = 'CRITICAL'
    elif score >= 50:   level = 'HIGH'
    elif score >=30:    level = 'MEDIUM'
    else:               level = 'LOW'

    return score, level     # return TWO values as a tuple

#Call the function
score,level = calculate_risk_score(0.42,True,5,47000)
print(f"Score:{score}, Level: {level}")
```

```
#DEFAULT PARAMETERS - Value used when caller doesnt provide one
def calculate_package(monthly_salary, bonus_pct=0.10,tax_rate=0.30):
  """Calculate the take-home salary. Default: 10% bonus, 30% tax."""
  annual = monthly_salary *12
  bonus = annual * bonus_pct
  tax = annual * tax_rate
  take_home = annual +bonus - tax
  return round(take_home/12,2)       # monthly take-home

#With defaults
print(calculate_package(65000))

#Override defaults
print(calculate_package(65000, bonus_pct=0.20))
print(calculate_package(65000,bonus_pct=0.20,tax_rate=0.25))  # custom both


#*args - accept ANY number of positional arguments
def team_salary_stats(*salaries):       #Create ANY number of positional arguments      #Explore key args *
  """Stats for any number of salaries."""
  if not salaries:
    return None
  return{
      "count": len(salaries),
      "total": sum(salaries),
      "average": round(sum(salaries)/len(salaries)),
      "highest": max(salaries),
      "lowest": min(salaries)
  }

stats = team_salary_stats(65000,82500,91000,47000,73000)
for key, val in stats.items():
  print(f" {key:<10}: ${val:,}" if key != "count" else f" {key:<10}: {val}")
```

```
#Exception Handling
#Real data is messy. Without error handling, one bad row crashes your ENTIRE program

#WITHOUT error handling - fragile
def bad_parser(value):
  return float(value)
#bad_parser("Luffy")   # error : could not convert string to float so thge lines below these would not get exectued
bad_parser(5)


#WITH error handling - robust
def safe_parse_salary(value, default=0.0):

    """Safely parse salary from any input."""
    try:
        cleaned = str(value).replace("$", "").replace(",", "").strip()
        return float(cleaned)
    except (ValueError):
        print(f"  Cannot parse '{value}' - using default {default}")
        return default
    except TypeError:
        print(f"Error: {e} -  using default {default}")
        return default

    except Exception as e:
      print(f" Error: {e} - using default")
      return default
```

```
#Test with messy data (this is what REAL data looks like)
test_values = ["65000","$82,500","N/A","","not available", None,"91000.50"]

print("Parsing messy salary data:")
for raw in test_values:
  result = safe_parse_salary(raw)
  print(f"Input: {str(raw):<20} -> ${result:,.2f}")
```
